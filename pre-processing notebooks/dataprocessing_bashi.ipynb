```python
import pandas
import os
```

```python
# DIRECTORIES
notebook_dir = os.getcwd()
base_dir = os.path.dirname(notebook_dir) # the notebook is in a subdirectory of the base directory
input_dir = os.path.join(base_dir, 'data', 'input')
print(f"Base directory: {base_dir}")
print(f"Input directory: {input_dir}")
```

```python
# FILES
drugdata_file = input_dir + '/bashi/cd-23-0388_table_s18_suppst18.xlsx'
drugdata_df = pandas.read_excel(drugdata_file)
print(drugdata_df.columns)

drugID_file = input_dir + '/bashi/drugnames_s2_suppst2.xlsx'
drugID_df = pandas.read_excel(drugID_file, sheet_name='Single drugs')
print(drugID_df.columns)
```

```python
# UNIQUE TRIPLETS IN DATASET
triplets = drugdata_df[['CELL_LINE_NAME', 'Compound1_name', 'Compound2_name']].copy()
triplets = triplets.drop_duplicates()
triplets = triplets.reset_index(drop=True)
print(f'Total unique (cell_line, drugA, drugB) triplets in dataset: {len(triplets)}')

# UNIQUE DRUG COMBINATIONS IN DATASET
drug_combinations = drugdata_df[['Compound1_name', 'Compound2_name']].copy()
drug_combinations = drug_combinations.drop_duplicates()
drug_combinations = drug_combinations.reset_index(drop=True)
print(f'Total unique drug combinations in dataset: {len(drug_combinations)}')
```

```python
# CLEAN CELL LINE NAMES
print(f'Total unique cell lines in dataset before cleaning: {drugdata_df["CELL_LINE_NAME"].nunique()}')

# Convert cell line names to strings and uppercase for consistency
drugdata_df['CELL_LINE_NAME'] = drugdata_df['CELL_LINE_NAME'].astype(str).str.upper()
print(f'Total unique cell lines in dataset after converting to uppercase: {drugdata_df["CELL_LINE_NAME"].nunique()}')

# Remove '-' or '+' from cell line names
drugdata_df['CELL_LINE_NAME'] = drugdata_df['CELL_LINE_NAME'].str.replace('-', '', regex=False)
drugdata_df['CELL_LINE_NAME'] = drugdata_df['CELL_LINE_NAME'].str.replace('+', '', regex=False)
print(f'Total unique cell lines in dataset after cleaning: {drugdata_df["CELL_LINE_NAME"].nunique()}')
# UNIQUE CELL LINES IN DATASET
clines = drugdata_df['CELL_LINE_NAME'].unique().tolist()
# clines.sort()
```

```python
# UNIQUE DRUGS IN DATASET
drug_names = drugdata_df[['Compound1_name', 'Compound2_name']].copy()
drug_names['Compound1_name'] = drug_names['Compound1_name'].str.upper()
drug_names['Compound2_name'] = drug_names['Compound2_name'].str.upper()

# Make list of total drugs in dataset
drug_list = sorted(set(drug_names['Compound1_name'].tolist() + drug_names['Compound2_name'].tolist()))
print(f'Total unique drugs in dataset: {len(drug_list)}')

# Save list as dataframe of single column 'drug_name'
drug_df = pandas.DataFrame(drug_list, columns=['drug_name'])
drugnames_file = input_dir + '/bashi/drug_names3.txt'
drug_df.to_csv(drugnames_file, index=False, header=True)
```

```python
# DOSES
# Change concentration units from uM to nM
drugdata_df['Compound1_maxc'] = pandas.to_numeric(drugdata_df['Compound1_maxc'], errors='coerce') * 1000
drugdata_df['Compound2_maxc'] = pandas.to_numeric(drugdata_df['Compound2_maxc'], errors='coerce') * 1000
```

```python
# UNIQUE TRIPLETS IN DATASET
triplets = drugdata_df[['CELL_LINE_NAME', 'drug_name_A', 'drug_name_B']].copy()
triplets = triplets.drop_duplicates()
triplets = triplets.reset_index(drop=True)
print(f'Total unique (cell_line, drugA, drugB) triplets in dataset: {len(triplets)}')
```

```python
# TISSUE INFORMATION
# Extract tissue information from cell line names
tissue_info = drugdata_df[['TISSUE', 'CELL_LINE_NAME']].copy()

# Manual tissue corrections
tissue_mapping = {
    'Haematopoietic and Lymphoid' : 'Lymphoid',
    'Head and Neck' : 'Head_neck',
    'Large Intestine' : 'Large_intestine',
    'Peripheral Nervous System': 'PNS',
    'Central Nervous System': 'CNS',
    'Biliary Tract': 'Biliary_tract',
}

tissue_info['TISSUE'] = tissue_info['TISSUE'].replace(tissue_mapping)
tissue_info['CELL_LINE_NAME'] = tissue_info['CELL_LINE_NAME'].str.upper()
tissue_info = tissue_info.drop_duplicates()
tissue_info = tissue_info.reset_index(drop=True)
tissue_info.to_csv(input_dir + '/bashi/tissue_cline3.csv', index=False, header=True)
```

```python
# SYNERGY DATA
synergies = drugdata_df[['TISSUE', 'CELL_LINE_NAME', 'Compound1_name', 'Compound1_target', 'Compound2_name', 'Compound2_target', 'Compound1_maxc', 'Compound2_maxc', 'Bliss_matrix']].copy()
rename_dict = {
    'CELL_LINE_NAME': 'cell_line',
    'TISSUE': 'tissue',
    'Compound1_name': 'drug_name_A',
    'Compound1_target': 'targets_A',
    'Compound2_name': 'drug_name_B',
    'Compound2_target': 'targets_B',
    'Compound1_maxc': 'conc_A',
    'Compound2_maxc': 'conc_B',
    'Bliss_matrix': 'synergy'
}

# Replace tissue names in synergy dataframe
synergies['TISSUE'] = synergies['TISSUE'].replace(tissue_mapping)

# Upper case drug names
synergies['Compound1_name'] = synergies['Compound1_name'].str.upper()
synergies['Compound2_name'] = synergies['Compound2_name'].str.upper()

# They measure excess inhibition: I = 1 - viability 
# Positive = synergy
# negative = antagonism
# synergies['Bliss_excess'] = -synergies['Bliss_excess']  # Flip sig if we want to measure excess viability instead of excess inhibition
synergies = synergies.rename(columns=rename_dict)
synergies.to_csv(input_dir + '/bashi/synergy_data3.csv', index=False, header=True)
```
